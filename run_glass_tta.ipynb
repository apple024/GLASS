### Train and evaluate with tta

```python
%%bash

datapath=./data/mvtec_ad
augpath=./data/dtd/images
classes=('capsule')
flags=($(for class in "${classes[@]}"; do echo '-d '"${class}"; done))

python main.py \
    --results_path ./results \
    --test ckpt \
  net \
    --tta \
    -b wideresnet50 \
    -le layer2 \
    -le layer3 \
    --meta_epochs 5 \
    --eval_epochs 1 \
    --limit 392 \
  dataset \
    --fg 1 \
    --batch_size 16 \
    "${flags[@]}" mvtec $datapath $augpath
```
